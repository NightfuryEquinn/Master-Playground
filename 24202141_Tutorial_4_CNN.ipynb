## Typical Use Cases

- Common in early layers extract low-level features, layers compress spatial info while increasing feature richness

```python
# Image is a gray scale (only one channel)
# Input size – 32, 32
# Output size – 5, 5

# With kernels and equal stride
m1 = nn.Conv2d(1, 6, 5)
m2 = nn.Conv2d(6, 6, 2, stride = 2)
m3 = nn.Conv2d(6, 16, 5)
m4 = nn.Conv2d(16, 16, 1, stride = 2)

input = torch.randn(1, 1, 32, 32)
hidden1 = m1(input)
hidden2 = m2(hidden1)
hidden3 = m3(hidden2)
output = m4(hidden3)

print('Input size: '+ str(np.shape(input)))
print('Hidden1 size: '+ str(np.shape(hidden1)))
print('Hidden2 size: '+ str(np.shape(hidden2)))
print('Hidden3 size: '+ str(np.shape(hidden3)))
print('Output size: '+ str(np.shape(output)))
```

## Typical Use Cases
- Embedded vision systems due to small network size
- Classic LeNet-5 architecture for image classification

```python
# Convolutions + stride reduce spatial dimensions while increasing depth (feature maps)
# Flatten converts 2D feature maps into 1D vector for fully connected layers
# FC layers perform classification or regression based on extracted features

# With kernels and equal stride
m1 = nn.Conv2d(1, 6, 5)
m2 = nn.Conv2d(6, 6, 2, stride = 2)
m3 = nn.Conv2d(6, 16, 5)
m4 = nn.Conv2d(16, 16, 1, stride = 2)
fc1 = nn.Linear(16 * 5 * 5, 120)
fc2 = nn.Linear(120, 84)
fc3 = nn.Linear(84, 10)

input = torch.randn(1, 1, 32, 32)
hidden1 = m1(input)
hidden2 = m2(hidden1)
hidden3 = m3(hidden2)
hidden4 = m4(hidden3)
fc1_layer = torch.flatten(hidden4, 1)
fc2_layer = fc1(fc1_layer)
fc3_layer = fc2(fc2_layer)
output = fc3(fc3_layer)

print('Input size: '+ str(np.shape(input)))
print('Hidden1 size: '+ str(np.shape(hidden1)))
print('Hidden2 size: '+ str(np.shape(hidden2)))
print('Hidden3 size: '+ str(np.shape(hidden3)))
print('Hidden4 size: '+ str(np.shape(hidden4)))
print('fc1_layer size: '+ str(np.shape(fc1_layer)))
print('fc2_layer size: '+ str(np.shape(fc2_layer)))
print('fc3_layer size: '+ str(np.shape(fc3_layer)))
print('Output size: '+ str(np.shape(output)))
```

```python
# Convolutional layers m1 - m4: Extract spatial features and reduce spatial dimensions
# Flatten: converts 16 x 5 x 5 feature maps to a 400-length vector for fully connected layers
# Fully connected layers fc1 - fc3: Perform classification (10-class output)

class ConvNet(torch.nn.Module):
  def __init__(self):
    super(ConvNet, self).__init__()
    self.m1 = nn.Conv2d(1, 6, 5)
    self.m2 = nn.Conv2d(6, 6, 2, stride = 2)
    self.m3 = nn.Conv2d(6, 16, 5)
    self.m4 = nn.Conv2d(16, 16, 1, stride = 2)
    self.fc1 = nn.Linear(16 * 5 * 5, 120)
    self.fc2 = nn.Linear(120, 84)
    self.fc3 = nn.Linear(84, 10)

  def forward(self, x):
    x = self.m1(x)
    x = self.m2(x)
    x = self.m3(x)
    x = self.m4(x)
    x = torch.flatten(x, 1)
    x = self.fc1(x)
    x = self.fc2(x)
    x = self.fc3(x)
    return x

# Useful for debugging shapes and confirming parameter counts
# Helps visualize how convolutional layers reduce spatial dimensions
from torchsummary import summary
model = ConvNet()
summary(model, (1, 32, 32))
```

## Typical Use Cases
- MNIST digit classification or any 28 x 28 grayscale image classification
- Standard CNN patterns: Convolutions + ReLU (Feature extraction) -> Max-pooling (Downsampling) -> Flatten (Classification)

```python
input = torch.randn(1, 1, 28, 28)

# With kernels and equal stride
m1 = nn.Conv2d(1, 32, kernel_size = 3, padding = "same", stride = 1)
m1_act = nn.ReLU()

hidden1 = m1(input)
hidden1 = m1_act(hidden1)
print('Input size: '+ str(np.shape(input)))
print('Hidden1 size: '+ str(np.shape(hidden1)))

m2 = nn.MaxPool2d(2, stride = 2)
hidden2 = m2(hidden1)
print('Hidden2 size: '+ str(np.shape(hidden2)))

m3 = nn.Conv2d(32, 64, kernel_size = 3, padding = "same", stride = 1)
m3_act = nn.ReLU()

hidden3 = m3(hidden2)
hidden3 = m3_act(hidden3)
print('Hidden3 size: '+ str(np.shape(hidden3)))

m4 = nn.MaxPool2d(2, stride=2)
hidden4 = m4(hidden3)
print('Hidden4 size: '+ str(np.shape(hidden4)))

fc1_layer = torch.flatten(hidden4, 1)
print('fc1_layer size: '+ str(np.shape(fc1_layer)))

fc2 = nn.Linear(np.shape(fc1_layer)[1], 128)
fc2_layer = fc2(fc1_layer)
print('fc2_layer size: '+ str(np.shape(fc2_layer)))

fc3 = nn.Linear(np.shape(fc2_layer)[1], 10)
output = fc3(fc2_layer)
print('Output size: '+ str(np.shape(output)))
```

```python
class ConvNet(torch.nn.Module):
  def __init__(self):
    super(ConvNet, self).__init__()
    self.m1 = nn.Conv2d(1, 32, kernel_size = 3, padding = "same", stride = 1)
    self.m1_act = nn.ReLU()
    self.m2 = nn.MaxPool2d(2, stride = 2)
    self.m3 = nn.Conv2d(32, 64, kernel_size = 3, padding = "same", stride = 1)
    self.m3_act = nn.ReLU()
    self.m4 = nn.MaxPool2d(2, stride = 2)
    self.fc2 = nn.Linear(3136, 128)
    self.fc3 = nn.Linear(128, 10)

  def forward(self, x):
    x = self.m1(x)
    x = self.m1_act(x)
    x = self.m2(x)
    x = self.m3(x)
    x = self.m3_act(x)
    x = self.m4(x)
    x = torch.flatten(x, 1)
    x = self.fc2(x)
    x = self.fc3(x)
    return x

from torchsummary import summary
model = ConvNet()
summary(model, (1, 28, 28))
```
